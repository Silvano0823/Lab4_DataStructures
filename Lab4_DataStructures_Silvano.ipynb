```python

# TASK 4: Identity-Based Computational Seed
LAST_NAME = "Silvano"
STUDENT_ID = "TUPM-26-7960"  # Replace with your actual student ID

seed_digit = int(STUDENT_ID[-1])
id_checksum = sum(int(d) for d in STUDENT_ID if d.isdigit())
vector_dim = len(LAST_NAME)

sys_config = {
    "auth_id": LAST_NAME,
    "base_seed": seed_digit,
    "checksum": id_checksum,
    "vector_dim": vector_dim,
    "status": "INITIALIZED"
}

print("System Configuration:")
print(sys_config)

# TASK 5: Dynamic List Operations
base_val = sys_config["base_seed"]
number_sequence = [
    base_val,
    base_val + 15,
    sys_config["checksum"]
]
print("Initial Sequence:", number_sequence)

number_sequence.append(base_val + 20)
print("Sequence After Append:", number_sequence)

# TASK 6: Sequential List Operations
new_numbers = [
    base_val + 5,
    sys_config["vector_dim"],
    base_val - 10
]
number_sequence.extend(new_numbers)
occurrences = number_sequence.count(base_val)
number_sequence.sort()

print("Sorted Sequence:", number_sequence)
print("Occurrences of Base Value:", occurrences)

# TASK 7: Tuples (Fixed Coordinates)
fixed_coords = (
    sys_config["vector_dim"],
    sys_config["base_seed"],
    0
)
x, y, z = fixed_coords

print("Fixed Coordinates:", fixed_coords)
print("X:", x)
print("Y:", y)
print("Z:", z)

try:
    fixed_coords[0] = 100
except TypeError:
    print("Error: Tuples cannot be modified!")

# TASK 8: Dictionary Key-Value Mapping
data_payload = {
    "identifier": sys_config["auth_id"],
    "dimensions": sys_config["vector_dim"],
    "status": sys_config["status"]
}
print("Original Payload:")

for key, value in data_payload.items():
    print(f"{key}: {value}")

data_payload["efficiency_rating"] = 98.5
print("Updated Payload:", data_payload)

# TASK 9: Set Deduplication
raw_data = [
    base_val,
    base_val + 15,
    sys_config["checksum"],
    base_val,
    base_val + 15,
    sys_config["checksum"]
]
unique_data = set(raw_data)

print("Original Data:", raw_data)
print("Unique Data:", unique_data)

# TASK 10: Mathematical Set Operations
reference_set = {
    base_val,
    base_val + 15,
    sys_config["checksum"]
}
intersection_set = unique_data.intersection(reference_set)
union_set = unique_data.union(reference_set)

print("Reference Set:", reference_set)
print("Intersection:", intersection_set)
print("Union:", union_set)

# TASK 11: Homogeneous Arrays
from array import array

int_array = array("i", [1, 2, 100])
print("Integer Array:", int_array)

try:
    int_array.append("hello")
except TypeError:
    print("Error: Only integers are allowed in this array!")

# TASK 12: List Comprehensions
squared_values = [x * base_val for x in range(1, 6)]
filtered_values = [
    value for value in squared_values if value > 15
]

print("Generated Values:", squared_values)
print("Filtered Values (> 15):", filtered_values)

# TASK 13: Double-Ended Queue
from collections import deque

task_queue = deque(number_sequence)
task_queue.append(100)
task_queue.appendleft(200)

print("Queue After Adding Elements:", task_queue)

task_queue.popleft()
print("Queue After Removing First Element:", task_queue)

# TASK 14: Fault-Tolerant Key Mapping
from collections import defaultdict

default_data = defaultdict(int)
default_data["active_key"] = sys_config["vector_dim"]

print("Existing Key:", default_data["active_key"])
print("Missing Key:", default_data["missing_key"])
```


## IV. Data and Observation

### Part 1: Functional Encapsulation
The dictionary `sys_config` stores related system information using named keys. This makes the configuration easier to read and maintain than a sequentially indexed list.

### Part 2: Dynamic List Operations
The `append()` method adds a new element to the end of a list. Python manages the required memory automatically, so the programmer does not need to manually reallocate memory.

### Part 3: Sequential Operations
The `extend()` method adds multiple elements to a list. The `count()` method counts occurrences of a value, while `sort()` arranges the values in ascending order.

### Part 4: Immutable Tuples
Tuples protect fixed coordinate values from accidental modification. Attempting to modify an element raises a `TypeError`.

### Part 5: Key-Value Mappings
Dictionaries associate keys with values, allowing direct access to information without searching through a sequential list.

### Part 6: Data De-duplication
Sets automatically remove duplicate values. This is useful when unique entries are needed, although set elements do not have a guaranteed display order.

### Part 7: Mathematical Set Operations
Intersection returns elements shared by both sets, while union combines their unique elements. These operations are useful for comparing collections of data.

### Part 8: Homogeneous Arrays
An integer array stores values of a specified numeric type. Attempting to append a string raises a `TypeError` because the value is incompatible with the array's type.

### Part 9: List Comprehensions
List comprehensions create lists concisely by applying an expression to each item in an iterable. A condition can filter the generated values.

### Part 10: Double-Ended Queues
A `deque` supports adding and removing elements at either end. The `appendleft()` and `popleft()` methods are useful for efficient queue operations.

### Part 11: Fault-Tolerant Key Mappings
A `defaultdict(int)` returns zero when a missing key is accessed. This avoids a `KeyError` and is useful when a default integer value is appropriate.



## V. Analysis and Interpretation

### 1. How does organizing data into specific structures (such as dictionaries, sets, and lists) affect data retrieval, program clarity, and system scalability?

Organizing data into specific structures makes programs easier to understand and manage. Lists store ordered items, dictionaries allow values to be retrieved using keys, and sets automatically remove duplicates. Choosing the appropriate data structure improves data retrieval and makes programs easier to maintain as the amount of data increases.

### 2. Why is it important to understand the distinction between mutable and immutable data types when designing programs that handle both dynamic inputs and fixed system parameters?

Mutable data types, such as lists and dictionaries, can be changed after creation. Immutable data types, such as tuples, cannot be modified after creation. Understanding this distinction helps programmers choose flexible structures for changing data and safer structures for fixed values, reducing accidental changes and errors.

### 3. How can utilizing sets to enforce uniqueness and perform mathematical operations (like unions or intersections) make programs more efficient at filtering and comparing large datasets?

Sets automatically remove duplicate values and provide operations such as union and intersection. These features make it easier to identify unique records, find common elements, and combine datasets without manually checking every duplicate.

### 4. In what situations does using specialized collections like double-ended queues (`deque`) optimize processing, and how do fault-tolerant structures (`defaultdict`) prevent runtime execution errors?

A `deque` is useful when a program frequently adds or removes elements at either end of a collection, such as in queues and scheduling systems. A `defaultdict` supplies a default value for missing keys, helping avoid `KeyError` exceptions when a key has not been assigned.
