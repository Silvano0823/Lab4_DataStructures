```python

# TASK 4: Identity-Based Computational Seed
LAST_NAME = "Silvano"
STUDENT_ID = "TUPM-26-7960"  # Replace with your actual student ID

seed_digit = int(STUDENT_ID[-1])
id_checksum = sum(int(d) for d in STUDENT_ID if d.isdigit())
vector_dim = len(LAST_NAME)

sys_config = {
    "auth_id": LAST_NAME,
    "base_seed": seed_digit,
    "checksum": id_checksum,
    "vector_dim": vector_dim,
    "status": "INITIALIZED"
}

print("System Configuration:")
print(sys_config)

# TASK 5: Dynamic List Operations
base_val = sys_config["base_seed"]
number_sequence = [
    base_val,
    base_val + 15,
    sys_config["checksum"]
]
print("Initial Sequence:", number_sequence)

number_sequence.append(base_val + 20)
print("Sequence After Append:", number_sequence)

# TASK 6: Sequential List Operations
new_numbers = [
    base_val + 5,
    sys_config["vector_dim"],
    base_val - 10
]
number_sequence.extend(new_numbers)
occurrences = number_sequence.count(base_val)
number_sequence.sort()

print("Sorted Sequence:", number_sequence)
print("Occurrences of Base Value:", occurrences)

# TASK 7: Tuples (Fixed Coordinates)
fixed_coords = (
    sys_config["vector_dim"],
    sys_config["base_seed"],
    0
)
x, y, z = fixed_coords

print("Fixed Coordinates:", fixed_coords)
print("X:", x)
print("Y:", y)
print("Z:", z)

try:
    fixed_coords[0] = 100
except TypeError:
    print("Error: Tuples cannot be modified!")

# TASK 8: Dictionary Key-Value Mapping
data_payload = {
    "identifier": sys_config["auth_id"],
    "dimensions": sys_config["vector_dim"],
    "status": sys_config["status"]
}
print("Original Payload:")

for key, value in data_payload.items():
    print(f"{key}: {value}")

data_payload["efficiency_rating"] = 98.5
print("Updated Payload:", data_payload)

# TASK 9: Set Deduplication
raw_data = [
    base_val,
    base_val + 15,
    sys_config["checksum"],
    base_val,
    base_val + 15,
    sys_config["checksum"]
]
unique_data = set(raw_data)

print("Original Data:", raw_data)
print("Unique Data:", unique_data)

# TASK 10: Mathematical Set Operations
reference_set = {
    base_val,
    base_val + 15,
    sys_config["checksum"]
}
intersection_set = unique_data.intersection(reference_set)
union_set = unique_data.union(reference_set)

print("Reference Set:", reference_set)
print("Intersection:", intersection_set)
print("Union:", union_set)

# TASK 11: Homogeneous Arrays
from array import array

int_array = array("i", [1, 2, 100])
print("Integer Array:", int_array)

try:
    int_array.append("hello")
except TypeError:
    print("Error: Only integers are allowed in this array!")

# TASK 12: List Comprehensions
squared_values = [x * base_val for x in range(1, 6)]
filtered_values = [
    value for value in squared_values if value > 15
]

print("Generated Values:", squared_values)
print("Filtered Values (> 15):", filtered_values)

# TASK 13: Double-Ended Queue
from collections import deque

task_queue = deque(number_sequence)
task_queue.append(100)
task_queue.appendleft(200)

print("Queue After Adding Elements:", task_queue)

task_queue.popleft()
print("Queue After Removing First Element:", task_queue)

# TASK 14: Fault-Tolerant Key Mapping
from collections import defaultdict

default_data = defaultdict(int)
default_data["active_key"] = sys_config["vector_dim"]

print("Existing Key:", default_data["active_key"])
print("Missing Key:", default_data["missing_key"])
```
