```python
# Student Information

LAST_NAME = "Silvano"
STUDENT_ID = "TUPM-26-7960"
SEED_NUM = int(STUDENT_ID[-1])

TRANSACTION_TYPES = [
    "DEPOSIT",
    "WITHDRAWAL",
    "TRANSFER",
    "PAYMENT",
    "DEPOSIT"
]

print("Student:", LAST_NAME)
print("Student ID:", STUDENT_ID)
print("Seed Number:", SEED_NUM)
print("Transaction Types:", TRANSACTION_TYPES)
```

```python
# Generate Transaction Records

transaction_records = []

for i in range(5):
    transaction = {
        "transaction_id": f"{LAST_NAME}_TXN_{i + 1}",
        "type": TRANSACTION_TYPES[i],
        "amount": (i + 1) * 100 + SEED_NUM
    }

    transaction_records.append(transaction)

print("Generated Transaction Records:")

for transaction in transaction_records:
    print(transaction)

print("\nTotal Transactions:", len(transaction_records))
```

```python
# Analyze Transaction History

total_amount = 0
transaction_counts = {}
transaction_totals = {}

for transaction in transaction_records:
    transaction_type = transaction["type"]
    amount = transaction["amount"]

    total_amount += amount

    transaction_counts[transaction_type] = (
        transaction_counts.get(transaction_type, 0) + 1
    )

    transaction_totals[transaction_type] = (
        transaction_totals.get(transaction_type, 0) + amount
    )

print("TRANSACTION HISTORY ANALYSIS")
print("=" * 35)

print("\nTotal Transactions:", len(transaction_records))
print("Total Transaction Amount:", total_amount)

print("\nTransaction Type Counts:")
for transaction_type, count in transaction_counts.items():
    print(transaction_type, ":", count)

print("\nTransaction Amounts by Type:")
for transaction_type, amount in transaction_totals.items():
    print(transaction_type, ":", amount)
```

```python
# Detect Repeated Transactions and Organize Records

seen_transactions = set()
repeated_transactions = []

for transaction in transaction_records:
    transaction_key = (
        transaction["transaction_id"],
        transaction["type"],
        transaction["amount"]
    )

    if transaction_key in seen_transactions:
        repeated_transactions.append(transaction)
    else:
        seen_transactions.add(transaction_key)

organized_transactions = sorted(
    transaction_records,
    key=lambda transaction: (
        transaction["type"],
        transaction["transaction_id"]
    )
)

print("ORGANIZED TRANSACTION HISTORY")
print("=" * 35)

for transaction in organized_transactions:
    print(transaction)

print("\nRepeated Transactions:", len(repeated_transactions))
print("Repeated Transaction Details:", repeated_transactions)
```

```python
# Final Transaction History Report

print("=" * 45)
print("TRANSACTION HISTORY ANALYZER")
print("Student:", LAST_NAME)
print("Student ID:", STUDENT_ID)
print("=" * 45)

print("\n1. Total Transactions:", len(transaction_records))
print("2. Total Transaction Amount:", total_amount)
print("3. Exact Duplicate Transactions:", len(repeated_transactions))

print("\n4. Transaction Type Summary")
for transaction_type, count in transaction_counts.items():
    print(
        transaction_type,
        "| Count:", count,
        "| Total Amount:", transaction_totals[transaction_type]
    )

print("\n5. Organized Transaction Records")
for transaction in organized_transactions:
    print(transaction)

print("\n6. Final Summary")
print("Transaction records analyzed successfully.")
print("Transactions were organized by type and ID.")
print("Exact duplicate transactions were checked.")
print("Transaction totals were calculated successfully.")

print("\nExecution completed successfully.")
```
